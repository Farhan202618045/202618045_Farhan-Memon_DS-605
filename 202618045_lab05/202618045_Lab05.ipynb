```python
def manual_one_hot_encode(train_df, test_df, categorical_cols):
    train_encoded = pd.DataFrame(index=train_df.index)
    test_encoded = pd.DataFrame(index=test_df.index)

    for col in categorical_cols:

        # Handle missing categorical values using training-data mode
        if train_df[col].isnull().any():
            mode_value = train_df[col].mode()[0]
            train_col = train_df[col].fillna(mode_value)
            test_col = test_df[col].fillna(mode_value)
        else:
            train_col = train_df[col]
            test_col = test_df[col]

        # Learn categories only from training data
        categories = sorted(train_col.unique())

        # Create one binary column for every category
        for category in categories:
            column_name = f"{col}_{category}"

            train_encoded[column_name] = (
                train_col == category
            ).astype(int)

            test_encoded[column_name] = (
                test_col == category
            ).astype(int)

    return train_encoded, test_encoded
```

```python
def manual_standard_scale(train_df, test_df, numerical_cols):
    train_scaled = pd.DataFrame(index=train_df.index)
    test_scaled = pd.DataFrame(index=test_df.index)

    means = {}
    stds = {}

    for col in numerical_cols:
        mean_value = train_df[col].mean()
        std_value = train_df[col].std(ddof=0)

        # Avoid division by zero if a feature has no variation
        if std_value == 0:
            std_value = 1.0

        means[col] = mean_value
        stds[col] = std_value

        train_scaled[col] = (
            train_df[col] - mean_value
        ) / std_value

        test_scaled[col] = (
            test_df[col] - mean_value
        ) / std_value

    return train_scaled, test_scaled, means, stds
```

```python
def linear_regression_closed_form(X, y):
    # Add a column of 1s for the intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Closed-form solution
    beta = np.linalg.pinv(X_with_intercept.T @ X_with_intercept) @ (
        X_with_intercept.T @ y
    )

    return beta
```

```python
def linear_regression_predict(X, beta):
    # Add a column of 1s for the intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Calculate predictions
    predictions = X_with_intercept @ beta

    return predictions
```

```python
def manual_mae(y_true, y_pred):
    errors = np.abs(y_true - y_pred)
    return np.mean(errors)
```

```python
def manual_rmse(y_true, y_pred):
    errors = y_true - y_pred
    squared_errors = errors ** 2
    return np.sqrt(np.mean(squared_errors))
```

```python
def manual_r2(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)

    return 1 - (ss_res / ss_tot)
```

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))
```

```python
def logistic_regression_train(X, y, learning_rate=0.05, n_iterations=5000):
    # Add intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Start all parameters at zero
    weights = np.zeros(X_with_intercept.shape[1])

    for _ in range(n_iterations):
        # Calculate raw scores
        z = X_with_intercept @ weights

        # Convert scores to probabilities
        probabilities = sigmoid(z)

        # Calculate gradient
        gradient = (
            X_with_intercept.T @ (probabilities - y)
        ) / len(y)

        # Update weights
        weights -= learning_rate * gradient

    return weights
```

```python
def logistic_predict_proba(X, weights):
    # Add intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Calculate raw scores
    z = X_with_intercept @ weights

    # Convert scores to probabilities
    probabilities = sigmoid(z)

    return probabilities
```

```python
def logistic_predict(X, weights, threshold=0.5):
    probabilities = logistic_predict_proba(X, weights)

    predictions = (probabilities >= threshold).astype(int)

    return predictions
```

## Part C - Optimization
### Optimized Manual Logistic Regression

```python
def logistic_regression_train_optimized(
    X,
    y,
    learning_rate=0.05,
    n_iterations=5000,
    tolerance=1e-7,
    patience=20
):
    # Add intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Initialize weights
    weights = np.zeros(X_with_intercept.shape[1])

    previous_loss = np.inf
    stable_iterations = 0

    for iteration in range(n_iterations):

        # Raw scores
        z = X_with_intercept @ weights

        # Predicted probabilities
        probabilities = sigmoid(z)

        # Gradient
        gradient = (
            X_with_intercept.T @ (probabilities - y)
        ) / len(y)

        # Update weights
        weights -= learning_rate * gradient

        # Calculate log loss
        probabilities_clipped = np.clip(
            probabilities,
            1e-15,
            1 - 1e-15
        )

        loss = -np.mean(
            y * np.log(probabilities_clipped)
            + (1 - y) * np.log(1 - probabilities_clipped)
        )

        # Check convergence
        if abs(previous_loss - loss) < tolerance:
            stable_iterations += 1

            if stable_iterations >= patience:
                break
        else:
            stable_iterations = 0

        previous_loss = loss

    return weights, iteration + 1
```

```python
start_time = time.perf_counter()

optimized_logistic_weights, optimized_iterations = (
    logistic_regression_train_optimized(
        X_manual_train_np,
        y_manual_clf_train_np,
        learning_rate=0.05,
        n_iterations=5000,
        tolerance=1e-7,
        patience=20
    )
)

optimized_classification_train_time = (
    time.perf_counter() - start_time
)

print(
    "Optimized training time:",
    optimized_classification_train_time,
    "seconds"
)

print(
    "Iterations actually used:",
    optimized_iterations
)
```

```python
def logistic_regression_train_fast(
    X,
    y,
    learning_rate=0.1,
    n_iterations=3000
):
    # Add intercept
    X_with_intercept = np.column_stack([
        np.ones(X.shape[0]),
        X
    ])

    # Precompute transpose because it never changes
    X_transpose = X_with_intercept.T

    # Initialize weights
    weights = np.zeros(X_with_intercept.shape[1])

    for _ in range(n_iterations):
        # Raw scores
        z = X_with_intercept @ weights

        # Probabilities
        probabilities = sigmoid(z)

        # Gradient
        gradient = (
            X_transpose @ (probabilities - y)
        ) / len(y)

        # Update weights
        weights -= learning_rate * gradient

    return weights
```
