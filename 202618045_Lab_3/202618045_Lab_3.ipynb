```python
# Task 2 drop (leakage + high-missingness)
df = df.drop(columns=['company', 'reservation_status', 'reservation_status_date'])

# Task 1 final step (now using cleaned columns)
X = df.drop(columns=['is_canceled'])
y = df['is_canceled']

numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = X.select_dtypes(include=['object', 'string']).columns.tolist()

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
```

### Task 2 - Missing Values, Leakage, and Outliers

```python
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_summary = pd.DataFrame({
    'missing_count': missing_count,
    'missing_pct': missing_pct
})
missing_summary = missing_summary[missing_summary['missing_count'] > 0].sort_values('missing_pct', ascending=False)
print(missing_summary)
```

```python
import matplotlib.pyplot as plt

cols_to_check = ['lead_time', 'adr', 'stays_in_weekend_nights', 'stays_in_week_nights',
                  'adults', 'children', 'babies', 'booking_changes', 'days_in_waiting_list']

fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()
for i, col in enumerate(cols_to_check):
    axes[i].boxplot(df[col].dropna())
    axes[i].set_title(col)
plt.tight_layout()
plt.show()
```

```python
def iqr_outlier_count(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return len(outliers), lower, upper

for col in cols_to_check:
    count, lower, upper = iqr_outlier_count(df, col)
    print(f"{col}: {count} outliers (bounds: {lower:.2f} to {upper:.2f})")
```

```python
before = df.shape[0]

# Fix known adr issues
df = df[(df['adr'] >= 0) & (df['adr'] < 5000)]

# Remove bookings with zero total guests
df = df[~((df['adults'] == 0) & (df['children'].fillna(0) == 0) & (df['babies'] == 0))]

after = df.shape[0]
print(f"Rows removed: {before - after}")
```

```python
X = df.drop(columns=['is_canceled'])
y = df['is_canceled']

numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = X.select_dtypes(include=['object', 'string']).columns.tolist()

print(X.shape)
print("Numerical:", numerical_cols)
print("Categorical:", categorical_cols)
```

### Task 3 - Create Two Preprocessing Pipelines

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder

# Numerical pipeline shared logic (impute, then scale)
numerical_pipeline_standard = Pipeline([
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler', StandardScaler())
])

numerical_pipeline_minmax = Pipeline([
    ('imputer', KNNImputer(n_neighbors=5)),
    ('scaler', MinMaxScaler())
])

# Categorical pipeline (same for both — impute + one-hot encode)
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Pipeline A: StandardScaler
preprocessor_A = ColumnTransformer([
    ('num', numerical_pipeline_standard, numerical_cols),
    ('cat', categorical_pipeline, categorical_cols)
])

# Pipeline B: MinMaxScaler
preprocessor_B = ColumnTransformer([
    ('num', numerical_pipeline_minmax, numerical_cols),
    ('cat', categorical_pipeline, categorical_cols)
])
```

# Part B - Model Training and Performance Evaluation

### Task 4 - Train Two Classification Models

```python
# Fit-transform each pipeline's preprocessor ONCE
X_train_A = preprocessor_A.fit_transform(X_train)
X_test_A = preprocessor_A.transform(X_test)

X_train_B = preprocessor_B.fit_transform(X_train)
X_test_B = preprocessor_B.transform(X_test)

print("Preprocessing done.")
```

```python
X_train_small = X_train.sample(3000, random_state=42)
y_train_small = y_train.loc[X_train_small.index]

import time
start = time.time()
X_train_A_small = preprocessor_A.fit_transform(X_train_small)
print("Sample preprocessing works, shape:", X_train_A_small.shape)
print("Time taken:", time.time() - start, "seconds")
```

```python
import time

start = time.time()
X_train_A = preprocessor_A.fit_transform(X_train)
X_test_A = preprocessor_A.transform(X_test)
elapsed_A = time.time() - start
print(f"Pipeline A done in {elapsed_A:.1f} seconds ({elapsed_A/60:.1f} minutes)")

start = time.time()
X_train_B = preprocessor_B.fit_transform(X_train)
X_test_B = preprocessor_B.transform(X_test)
elapsed_B = time.time() - start
print(f"Pipeline B done in {elapsed_B:.1f} seconds ({elapsed_B/60:.1f} minutes)")
```

```python
import numpy as np
np.save('X_train_A.npy', X_train_A)
np.save('X_test_A.npy', X_test_A)
np.save('X_train_B.npy', X_train_B)
np.save('X_test_B.npy', X_test_B)
print("Saved preprocessed arrays.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

model_lr_A = LogisticRegression(max_iter=1000)
model_lr_A.fit(X_train_A, y_train)

model_lr_B = LogisticRegression(max_iter=1000)
model_lr_B.fit(X_train_B, y_train)

model_dt_A = DecisionTreeClassifier(random_state=42)
model_dt_A.fit(X_train_A, y_train)

model_dt_B = DecisionTreeClassifier(random_state=42)
model_dt_B.fit(X_train_B, y_train)

print("All 4 models trained.")
```

### Task 5 - Evaluate and Compare the Four Results

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_model(model, X_train, y_train, X_test, y_test, name):
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)
    
    train_acc = accuracy_score(y_train, train_pred)
    test_acc = accuracy_score(y_test, test_pred)
    precision = precision_score(y_test, test_pred)
    recall = recall_score(y_test, test_pred)
    f1 = f1_score(y_test, test_pred)
    
    return {
        'Model-Pipeline': name,
        'Train Accuracy': train_acc,
        'Test Accuracy': test_acc,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1
    }

results = []
results.append(evaluate_model(model_lr_A, X_train_A, y_train, X_test_A, y_test, 'LogReg + Pipeline A (Standard)'))
results.append(evaluate_model(model_lr_B, X_train_B, y_train, X_test_B, y_test, 'LogReg + Pipeline B (MinMax)'))
results.append(evaluate_model(model_dt_A, X_train_A, y_train, X_test_A, y_test, 'DecisionTree + Pipeline A (Standard)'))
results.append(evaluate_model(model_dt_B, X_train_B, y_train, X_test_B, y_test, 'DecisionTree + Pipeline B (MinMax)'))

import pandas as pd
comparison_table = pd.DataFrame(results)
print(comparison_table.to_string(index=False))
```

```python
comparison_table.to_csv('comparison_table.csv', index=False)
print("Saved comparison_table.csv")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Best Logistic Regression (Pipeline A)
cm_lr = confusion_matrix(y_test, model_lr_A.predict(X_test_A))
disp_lr = ConfusionMatrixDisplay(confusion_matrix=cm_lr, display_labels=['Not Canceled', 'Canceled'])
disp_lr.plot(ax=axes[0], cmap='Blues', colorbar=False)
axes[0].set_title('Best Logistic Regression (Pipeline A)')

# Best Decision Tree (Pipeline A)
cm_dt = confusion_matrix(y_test, model_dt_A.predict(X_test_A))
disp_dt = ConfusionMatrixDisplay(confusion_matrix=cm_dt, display_labels=['Not Canceled', 'Canceled'])
disp_dt.plot(ax=axes[1], cmap='Blues', colorbar=False)
axes[1].set_title('Best Decision Tree (Pipeline A)')

plt.tight_layout()
plt.savefig('confusion_matrices.png', dpi=150)
plt.show()
```

### Task 6 - Final Observations

##### 1. Which preprocessing-model combination gives the best overall result?
Decision Tree with either pipeline gives the highest raw test accuracy (~85.9%), edging out Logistic Regression (~81.6%). However, this comes with a caveat — the Decision Tree shows a large train-test gap (99.6% vs 85.9%), indicating overfitting, so its test performance is likely somewhat optimistic compared to how it would perform on genuinely new, unseen data. Logistic Regression, while lower in raw accuracy, generalizes more reliably (train/test gap of only ~0.2%).

##### 2. Does StandardScaler or MinMaxScaler affect Logistic Regression more?
StandardScaler (Pipeline A) performed marginally better than MinMaxScaler (Pipeline B) — 81.6% vs 81.3% test accuracy, with similar small edges in precision and recall. The difference is modest but consistent, which fits the theory that StandardScaler's centered, unit-variance output tends to suit gradient-based optimizers slightly better than MinMax's fixed 0-1 range.

##### 3. Does scaling make a major difference for the Decision Tree?
No — test accuracy was virtually identical (85.9324% vs 85.9282%), a difference in the third decimal place. This is expected: Decision Trees split on per-feature thresholds rather than computing distances or gradients, so the specific scaling method doesn't change where those thresholds fall.

##### 4. Write 4-5 short observations supported by the performance table and confusion matrices.
Decision Tree shows clear overfitting (train accuracy 99.6% vs test accuracy 85.9%) since it was trained with no depth restriction, allowing it to grow until it nearly memorized the training data. Logistic Regression shows no meaningful overfitting.
Logistic Regression has higher precision (~81%) but lower recall (~66%) than Decision Tree (~80.5% precision, ~81.8% recall) — meaning Logistic Regression is more conservative (fewer false alarms) while Decision Tree catches more actual cancellations at a similar precision cost.
