```python
# Two matrices
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

# Addition
addition = M1 + M2
print("Addition:\n", addition)

# Element-wise multiplication
elementwise = M1 * M2
print("\nElement-wise multiplication:\n", elementwise)

# Matrix multiplication
matmul = M1 @ M2  # or np.matmul(M1, M2)
print("\nMatrix multiplication:\n", matmul)
```

```python
# Using M1 as our square matrix
transpose = M1.T
determinant = np.linalg.det(M1)

print("Transpose:\n", transpose)
print("\nDeterminant:", determinant)
```

```python
if determinant != 0:
    inverse = np.linalg.inv(M1)
    print("Inverse:\n", inverse)

    # Verify M1 @ M1_inverse ≈ Identity matrix
    identity_check = M1 @ inverse
    print("\nM1 @ M1_inverse:\n", identity_check)

    is_identity = np.allclose(identity_check, np.eye(2))
    print("\nIs M1 @ M1_inverse ≈ Identity?", is_identity)
else:
    print("Matrix is singular (determinant = 0), no inverse exists.")
```

## Task 2 – Observations

- Matrix addition and element-wise multiplication both operate position-by-position, 
  while matrix multiplication (`@`) follows the standard row-by-column dot product rule — 
  the results differ significantly from element-wise multiplication.
- The transpose of M1 swaps its rows and columns, which is a purely structural change 
  with no effect on the underlying values.
- The determinant of M1 was non-zero, confirming the matrix is invertible (a singular 
  matrix would have determinant = 0 and no inverse).
- Multiplying M1 by its inverse produced a matrix extremely close to the identity matrix, 
  and `np.allclose()` confirmed this holds within floating-point tolerance (exact equality 
  isn't guaranteed due to floating-point rounding, which is why `allclose` rather than `==` 
  is the correct check).
- All operations were performed using NumPy's vectorized functions/operators rather than 
  explicit loops, which is both more concise and significantly faster for numerical computation.

**Difference between loc and iloc:**
`loc` selects data using labels — row index labels and column names — and the end of a slice 
is *inclusive* (e.g., `0:4` includes row 4). 
`iloc` selects data using integer positions only, regardless of labels, and the end of a slice 
is *exclusive* (e.g., `0:5` includes positions 0-4, not 5).

```python
print("Missing Age values BEFORE filling:", df['Age'].isnull().sum())

df['Age_filled_mean'] = df['Age'].fillna(df['Age'].mean())

print("Missing Age values AFTER filling:", df['Age_filled_mean'].isnull().sum())
```

```python
# Mean imputation
age_mean_imp = df['Age'].fillna(df['Age'].mean())

# Median imputation
age_median_imp = df['Age'].fillna(df['Age'].median())

# Mode imputation
age_mode_imp = df['Age'].fillna(df['Age'].mode()[0])

# Random value imputation (randomly sampled from existing non-null values)
np.random.seed(42)
age_random_imp = df['Age'].copy()
null_indices = age_random_imp[age_random_imp.isnull()].index
random_samples = np.random.choice(df['Age'].dropna().values, size=len(null_indices))
age_random_imp.loc[null_indices] = random_samples

print("Original mean Age (excluding NaN):", df['Age'].mean())
print("Mean imputation - resulting mean:", age_mean_imp.mean())
print("Median imputation - resulting mean:", age_median_imp.mean())
print("Mode imputation - resulting mean:", age_mode_imp.mean())
print("Random imputation - resulting mean:", age_random_imp.mean())
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower bound: {lower_bound}")
print(f"Upper bound: {upper_bound}")
print(f"Number of Fare outliers: {len(fare_outliers)}")
```
